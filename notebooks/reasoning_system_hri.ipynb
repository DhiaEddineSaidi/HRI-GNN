```python
from torch.utils.data import DataLoader
from hri_gnn.datasets.rule_dataset import RuleDetectorDataset, rule_detector_collate, load_rule_label_cache
from hri_gnn.utils import paths

output_path = paths.get_output_path()
paths.output_paths(output_path)

dataset_name = 'raven-20000'
dataset_path = paths.get_dataset_path(dataset_name)[0]
data_path = dataset_path.parent.parent
rule_label_cache = data_path / 'processed' / 'rule_labels_all.json'
percentage_to_use: float = 0.75
batch_size: int = 256

train_ds = RuleDetectorDataset(dataset_name, 'train', rule_label_cache, percentage_to_use)
test_ds = RuleDetectorDataset(dataset_name, 'test', rule_label_cache, percentage_to_use)
val_ds   = RuleDetectorDataset(dataset_name, 'val',   rule_label_cache, percentage_to_use)

loader_kw = dict(collate_fn=rule_detector_collate, num_workers=4, persistent_workers=True, pin_memory=True)

train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True,  **loader_kw)
test_loader = DataLoader(test_ds, batch_size=batch_size, shuffle=False,  **loader_kw)
val_loader   = DataLoader(val_ds,   batch_size=batch_size, shuffle=False, **loader_kw)
```

```python
from itertools import islice

batches = list(islice(train_loader,100))
len(batches)

```

```python
# load rule_label_cache
from hri_gnn.datasets.rule_dataset import  load_rule_label_cache

all_label= load_rule_label_cache(rule_label_cache)
```
